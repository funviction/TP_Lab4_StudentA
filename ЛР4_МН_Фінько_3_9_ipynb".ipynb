## 6. Аналіз та обробка викидів

Для неперервних числових ознак навчальної вибірки:
- визначити потенційні викиди методом IQR;
- вивести кількість потенційних викидів для кожної ознаки;
- обґрунтувати спосіб їх обробки.

Для невеликого набору даних не видаляти автоматично всі рядки з викидами. Використати обмеження крайніх значень межами, визначеними за навчальною вибіркою.

Цільову змінну `price` не використовувати для визначення меж викидів.  
До тестової вибірки застосувати ті самі межі, які були отримані за train.


```python
DISC_COLS = ['symboling']
cont_cols = [col for col in tr_X.select_dtypes(include='number').columns
                 if col not in DISC_COLS]
print('Неперервні числові ознаки:', cont_cols)


class OutlierClipper(BaseEstimator, TransformerMixin):
    """Обмежує значення межами [Q1 - k*IQR, Q3 + k*IQR], обчисленими під час fit (тобто за train)."""

    def __init__(self, mult=1.5):
        self.mult = mult

    def fit(self, X_in, y_in=None):
        df_x = pd.DataFrame(X_in)
        perc_25, perc_75 = df_x.quantile(0.25), df_x.quantile(0.75)
        range_iqr = perc_75 - perc_25
        self.min_val_ = (perc_25 - self.mult * range_iqr).to_numpy()
        self.max_val_ = (perc_75 + self.mult * range_iqr).to_numpy()
        self.input_features_ = np.asarray(df_x.columns, dtype=object)
        return self

    def transform(self, X_in):
        df_x = pd.DataFrame(X_in).copy()
        return df_x.clip(lower=self.min_val_, upper=self.max_val_, axis=1)

    def get_feature_names_out(self, input_features=None):
        return np.asarray(self.input_features_ if input_features is None else input_features, dtype=object)


iqr_transformer = OutlierClipper(mult=1.5).fit(tr_X[cont_cols])
limit_bounds = pd.DataFrame({'Нижня межа': iqr_transformer.min_val_, 'Верхня межа': iqr_transformer.max_val_}, index=cont_cols)


def calc_outliers(data_part):
    return ((data_part[cont_cols] < limit_bounds['Нижня межа']) | (data_part[cont_cols] > limit_bounds['Верхня межа'])).sum()


outliers_stats = limit_bounds.assign(
    **{'Викидів у train': calc_outliers(tr_X),
       '% у train': (calc_outliers(tr_X) / len(tr_X) * 100).round(1),
       'Викидів у test (межі train)': calc_outliers(te_X)}
).sort_values('Викидів у train', ascending=False)
display(outliers_stats)
bad_rows_count = ((tr_X[cont_cols] < limit_bounds['Нижня межа']) |
                      (tr_X[cont_cols] > limit_bounds['Верхня межа'])).any(axis=1).sum()
print(f'Рядків train, що містять хоча б один викид: {bad_rows_count} з {len(tr_X)} '
      f'({bad_rows_count / len(tr_X):.0%}) — видаляти їх недоцільно.')

tr_X_clipped = tr_X.copy()
te_X_clipped = te_X.copy()
tr_X_clipped[cont_cols] = iqr_transformer.transform(tr_X[cont_cols]).to_numpy()
te_X_clipped[cont_cols] = iqr_transformer.transform(te_X[cont_cols]).to_numpy()
print('Викидів у train після обмеження:', int(calc_outliers(tr_X_clipped).sum()),
      '| у test після обмеження:', int(calc_outliers(te_X_clipped).sum()))

cols_with_outs = outliers_stats.index[outliers_stats['Викидів у train'] > 0].tolist()
fig_out, axes_out = plt.subplots(2, len(cols_with_outs), figsize=(2.6 * len(cols_with_outs), 6))
for idx, c_name in enumerate(cols_with_outs):
    sns.boxplot(y=tr_X[c_name], ax=axes_out[0, idx], color='tab:orange')
    sns.boxplot(y=tr_X_clipped[c_name], ax=axes_out[1, idx], color='tab:green')
    axes_out[0, idx].set_title(c_name)
    axes_out[0, idx].set_ylabel('до' if idx == 0 else '')
    axes_out[1, idx].set_ylabel('після' if idx == 0 else '')
plt.suptitle('Train: ознаки з викидами до (вгорі) і після (внизу) обмеження межами IQR', fontsize=13)
plt.tight_layout()
plt.show()
```

За правилом IQR (межі Q1 − 1.5·IQR, Q3 + 1.5·IQR, обчислені тільки за train, без price) потенційні викиди мають 9 з 13 неперервних ознак. Найбільше у compressionratio — 20 (12.2%); це дизельні авто зі ступенем стиснення 21–23, тобто природна група. Інформацію про тип палива зберігає ознака fueltype. У horsepower і enginesize — це реальні потужні дорогі автомобілі. Хоча б один викид містять 50 із 164 рядків train (30%). Видалення таких рядків скоротило б вибірку майже на третину і прибрало б саме дорогі авто. Тому викиди не видаляються, а обмежуються межами train (clipping / вінзоризація) за допомогою кастомного класу OutlierClipper. Так зберігаються всі спостереження, а вплив екстремальних значень на лінійні моделі та SVR зменшується. До test застосовано ті самі межі. Цей крок включено першим етапом передобробки в Pipeline.

## 7. Аналіз кореляцій та відбір ознак

За навчальною вибіркою:
- побудувати матрицю кореляцій числових ознак;
- знайти пари ознак із сильною кореляцією, для яких `|r| > 0.85`;
- для кожної такої пари залишити одну ознаку;
- під час вибору врахувати силу зв'язку кожної ознаки з цільовою змінною `price`;
- вивести список сильно корельованих пар та перелік ознак, які видаляються.

Із тестової вибірки видалити ті самі ознаки.  
Не використовувати тестову вибірку для прийняття рішення про відбір ознак.


```python
R_THRESH = 0.85
num_vars = cont_cols + DISC_COLS
df_corr = tr_X_clipped[num_vars].assign(price=tr_y)
mat_corr = df_corr.corr()

plt.figure(figsize=(13, 10))
upper_mask = np.triu(np.ones_like(mat_corr, dtype=bool), k=1)
sns.heatmap(mat_corr, mask=upper_mask, annot=True, fmt='.2f', cmap='coolwarm', center=0,
            vmin=-1, vmax=1, annot_kws={'size': 8}, square=True, cbar_kws={'shrink': 0.8})
plt.title('Матриця кореляцій числових ознак (train)')
plt.tight_layout()
plt.show()

corr_target = mat_corr[TARGET_COL].drop(TARGET_COL).abs()

high_corr_pairs = []
for idx_i, feat1 in enumerate(num_vars):
    for feat2 in num_vars[idx_i + 1:]:
        corr_val = mat_corr.loc[feat1, feat2]
        if abs(corr_val) > R_THRESH:
            high_corr_pairs.append((feat1, feat2, corr_val))
high_corr_pairs.sort(key=lambda p: abs(p[2]), reverse=True)

drop_set, report_rows = set(), []
for feat1, feat2, corr_val in high_corr_pairs:
    if feat1 in drop_set or feat2 in drop_set:
        action = 'пара вже розв\'язана (одна з ознак видалена раніше)'
    else:
        drop_feat = feat1 if corr_target[feat1] < corr_target[feat2] else feat2
        drop_set.add(drop_feat)
        action = f'видалити {drop_feat}'
    report_rows.append({'Ознака 1': feat1, 'Ознака 2': feat2, 'r': corr_val,
                        '|r(ознака 1, price)|': corr_target[feat1],
                        '|r(ознака 2, price)|': corr_target[feat2], 'Рішення': action})

print(f'Сильно корельовані пари (|r| > {R_THRESH}):')
display(pd.DataFrame(report_rows))
drop_list = sorted(drop_set)
print('Ознаки, які видаляються:', drop_list)

tr_X = tr_X.drop(columns=drop_list)
te_X = te_X.drop(columns=drop_list)
tr_X_clipped = tr_X_clipped.drop(columns=drop_list)
te_X_clipped = te_X_clipped.drop(columns=drop_list)
cont_cols = [c for c in cont_cols if c not in drop_list]
print('tr_X:', tr_X.shape, '| te_X:', te_X.shape)
print('Неперервні ознаки, що залишились:', cont_cols)
```

У train знайдено 4 пари ознак з |r| > 0.85. Пари оброблялись від найсильнішої кореляції, і в кожній залишалась ознака, тісніше пов'язана з ціною (price):
Видалено citympg, wheelbase, carlength, carwidth: вони дублюють інформацію про розмір і витрату пального, яку зберігають curbweight (найсильніший зв'язок з ціною серед розмірних ознак, r = 0.82) і highwaympg. Пара enginesize–curbweight (r = 0.849) поріг не перевищує, тому обидві ознаки залишено. Із test видалено ті самі 4 ознаки. Залишилось 20 ознак.

## 8. Підготовка числових і категоріальних ознак

Розділити ознаки на:
- неперервні числові;
- дискретну числову ознаку `symboling`;
- категоріальні.

Для неперервних числових ознак виконати послідовно:
1. перетворення Yeo–Johnson для зменшення асиметрії;
2. масштабування за допомогою StandardScaler.

Для `symboling` виконати масштабування без Yeo–Johnson.

Категоріальні ознаки закодувати методом One-Hot Encoding. Передбачити коректну обробку категорій у test, яких не було у train.

Усі перетворення об'єднати в єдиний механізм передобробки даних, який навчається тільки на `X_train`.


```python
cat_cols = tr_X.select_dtypes(exclude='number').columns.tolist()
print('Неперервні числові ознаки:', cont_cols)
print('Дискретна числова ознака:', DISC_COLS)
print('Категоріальні ознаки:', cat_cols)

pipe_cont = Pipeline([
    ('clip', OutlierClipper(mult=1.5)),
    ('yeo_johnson', PowerTransformer(method='yeo-johnson', standardize=False)),
    ('scaler', StandardScaler()),
])
pipe_disc = Pipeline([('scaler', StandardScaler())])
pipe_cat = Pipeline([
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False)),
])

data_prep = ColumnTransformer([
    ('cont', pipe_cont, cont_cols),
    ('disc', pipe_disc, DISC_COLS),
    ('cat', pipe_cat, cat_cols),
], remainder='drop')

tr_X_ready = data_prep.fit_transform(tr_X)
te_X_ready = data_prep.transform(te_X)
out_features = data_prep.get_feature_names_out()
print('\nРозмірність після передобробки: train', tr_X_ready.shape, '| test', te_X_ready.shape)
print('Кількість ознак після One-Hot Encoding:', len(out_features))

new_cats = {c: sorted(set(te_X[c]) - set(tr_X[c])) for c in cat_cols}
new_cats = {c: v for c, v in new_cats.items() if v}
print('Категорії test, яких немає у train:', new_cats if new_cats else 'немає')
for c_name, vals in new_cats.items():
    ohe_indices = [idx for idx, n in enumerate(out_features) if n.startswith(f'cat__{c_name}_')]
    row_flags = te_X[c_name].isin(vals).to_numpy()
    print(f'  {c_name}={vals}: рядків у test — {row_flags.sum()}, '
          f'сума їх one-hot значень по {c_name} = {te_X_ready[row_flags][:, ohe_indices].sum():.0f} '
          f'(закодовано нулями, помилки немає)')
```

Ознаки розділено на три групи, кожна обробляється у своїй гілці ColumnTransformer (змінна data_prep):

9 неперервних: обмеження викидів (OutlierClipper) → Yeo–Johnson → StandardScaler;

symboling: лише StandardScaler (дискретна порядкова шкала);

10 категоріальних: One-Hot Encoding з handle_unknown='ignore'.

Після передобробки виходить 69 ознак. У test трапилась категорія cylindernumber = 'three', якої немає в train. Вона закодована нулями, тож помилки не виникає. Увесь механізм навчається лише на tr_X (fit), а test тільки трансформується (transform).

## 9. Перевірка трансформації числових ознак

Перевірити результат трансформації неперервних числових ознак:
- застосувати до train ту саму послідовність перетворень, яка буде використана в моделях;
- повторно обчислити коефіцієнти асиметрії;
- порівняти їх із початковими значеннями;
- зробити короткий висновок щодо ефективності трансформації.


```python
cont_step = data_prep.named_transformers_['cont']
df_tr_cont = pd.DataFrame(cont_step.transform(tr_X[cont_cols]),
                                        columns=cont_cols, index=tr_X.index)

df_skew_comp = pd.DataFrame({
    'skew до': tr_X[cont_cols].skew(),
    'skew після обмеження викидів': tr_X_clipped[cont_cols].skew(),
    'skew після clip + Yeo–Johnson + StandardScaler': df_tr_cont.skew(),
})
df_skew_comp['|skew| зменшився'] = (df_skew_comp.iloc[:, 2].abs() < df_skew_comp.iloc[:, 0].abs())
df_skew_comp = df_skew_comp.sort_values('skew до', key=np.abs, ascending=False)
display(df_skew_comp)
print('Середній |skew| до:', round(df_skew_comp['skew до'].abs().mean(), 3),
      '| після:', round(df_skew_comp.iloc[:, 2].abs().mean(), 3))
print('Ознак з |skew| > 1 до:', int((df_skew_comp['skew до'].abs() > 1).sum()),
      '| після:', int((df_skew_comp.iloc[:, 2].abs() > 1).sum()))
print('Після StandardScaler: середні ≈', round(float(df_tr_cont.mean().abs().max()), 6),
      ', стандартні відхилення ≈', df_tr_cont.std(ddof=0).round(3).unique())

lambda_vals = pd.Series(cont_step.named_steps['yeo_johnson'].lambdas_, index=cont_cols, name='λ Yeo–Johnson')
display(lambda_vals.to_frame().T)

df_skew_comp.iloc[:, [0, 2]].plot(kind='bar', figsize=(12, 4.5), color=['tab:orange', 'tab:green'])
plt.axhline(0, color='black', lw=0.8)
plt.axhline(0.5, color='gray', ls='--', lw=0.8)
plt.axhline(-0.5, color='gray', ls='--', lw=0.8)
plt.title('Коефіцієнти асиметрії неперервних ознак train до і після трансформації')
plt.ylabel('skew')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
```

Трансформація неперервних ознак виявилась ефективною. Середній |skew| знизився з 0.915 до 0.023, а кількість ознак із сильною асиметрією (|skew| > 1) — з 3 до 0. Найбільше змінились compressionratio (2.74 → 0.005), enginesize (2.05 → 0.045) та horsepower (1.55 → 0.066). Майже симетричними розподіли робить саме Yeo–Johnson, а StandardScaler приводить ознаки до середнього 0 і стандартного відхилення 1.

## 10. Побудова регресійних моделей

Побудувати та навчити такі моделі:
- Linear Regression;
- Ridge Regression;
- Lasso Regression;
- Random Forest Regressor;
- Gradient Boosting Regressor;
- Support Vector Regressor (SVR).

Для кожної моделі створити єдиний Pipeline, який містить:
1. попередню обробку ознак;
2. відповідну регресійну модель.

Використати однакову схему підготовки даних для коректного порівняння моделей.


```python
from sklearn.base import clone

models_dict = {
    'Linear Regression': LinearRegression(),
    'Ridge Regression': Ridge(alpha=1.0),
    'Lasso Regression': Lasso(alpha=10.0, max_iter=100_000),
    'Random Forest': RandomForestRegressor(n_estimators=300, random_state=RAND_SEED, n_jobs=-1),
    'Gradient Boosting': GradientBoostingRegressor(n_estimators=300, learning_rate=0.05, max_depth=3,
                                                   random_state=RAND_SEED),
    'SVR': TransformedTargetRegressor(regressor=SVR(kernel='rbf', C=10, epsilon=0.05),
                                      transformer=StandardScaler()),
}

pipes_dict = {m_name: Pipeline([('preprocessing', clone(data_prep)), ('model', mod)])
              for m_name, mod in models_dict.items()}

for m_name, current_pipe in pipes_dict.items():
    current_pipe.fit(tr_X, tr_y)
    print(f'{m_name:<18} навчено')

pipes_dict['Random Forest']
```

Побудовано 6 Pipeline з однаковою передобробкою (data_prep) і різними моделями: LinearRegression, Ridge, Lasso, RandomForest, GradientBoosting, SVR.
Для SVR ціну додатково стандартизовано через TransformedTargetRegressor, бо ε задається в одиницях цільової змінної, і без масштабування price модель практично не навчається. Для ансамблів зафіксовано RAND_SEED=42.

## 11. Оцінювання якості моделей

Для кожної моделі:
- виконати навчання на тренувальній вибірці;
- отримати прогноз для тестової вибірки;
- обчислити метрики **R², MAE, MSE та MAPE**.

`MAPE` подати у відсотках.

Результати всіх моделей об'єднати в одну порівняльну таблицю та впорядкувати за значенням R² від найбільшого до найменшого.

Під час інтерпретації врахувати:
- R² — чим більше, тим краще;
- MAE, MSE, MAPE — чим менше, тим краще.


```python
cv_split = KFold(n_splits=5, shuffle=True, random_state=RAND_SEED)
metrics_list, preds_dict = [], {}

for m_name, current_pipe in pipes_dict.items():
    preds = current_pipe.predict(te_X)
    preds_dict[m_name] = preds
    r2_cv_scores = cross_val_score(clone(current_pipe), tr_X, tr_y, cv=cv_split, scoring='r2')
    metrics_list.append({
        'Модель': m_name,
        'R²': r2_score(te_y, preds),
        'MAE': mean_absolute_error(te_y, preds),
        'MSE': mean_squared_error(te_y, preds),
        'MAPE, %': mean_absolute_percentage_error(te_y, preds) * 100,
        'RMSE': np.sqrt(mean_squared_error(te_y, preds)),
        'R² train': r2_score(tr_y, current_pipe.predict(tr_X)),
        'R² CV (5 fold, train)': r2_cv_scores.mean(),
    })

df_metrics = (pd.DataFrame(metrics_list).sort_values('R²', ascending=False)
              .reset_index(drop=True))
df_metrics.index += 1
df_metrics['Різниця R² train − test'] = df_metrics['R² train'] - df_metrics['R²']

print('Метрики на тестовій вибірці (впорядковано за R²). Основні: R², MAE, MSE, MAPE;')
print('додатково: RMSE і R² на train / крос-валідації для аналізу перенавчання.')
display(df_metrics.style.format({
    'R²': '{:.4f}', 'MAE': '{:,.0f}', 'MSE': '{:,.0f}', 'MAPE, %': '{:.2f}', 'RMSE': '{:,.0f}',
    'R² train': '{:.4f}', 'R² CV (5 fold, train)': '{:.4f}', 'Різниця R² train − test': '{:.4f}'
}).highlight_max(subset=['R²'], color='lightgreen')
  .highlight_min(subset=['MAE', 'MSE', 'MAPE, %'], color='lightgreen'))

fig_metrics, axes_metrics = plt.subplots(1, 3, figsize=(17, 4))
sort_order = df_metrics['Модель']
sns.barplot(data=df_metrics, y='Модель', x='R²', ax=axes_metrics[0], order=sort_order, color='tab:blue')
sns.barplot(data=df_metrics, y='Модель', x='MAE', ax=axes_metrics[1], order=sort_order, color='tab:orange')
sns.barplot(data=df_metrics, y='Модель', x='MAPE, %', ax=axes_metrics[2], order=sort_order, color='tab:green')
axes_metrics[0].set_xlim(min(0.5, df_metrics['R²'].min() - 0.05), 1)
for axis in axes_metrics[1:]:
    axis.set_ylabel('')
plt.suptitle('Порівняння моделей на тестовій вибірці')
plt.tight_layout()
plt.show()
```

Найкращий результат за всіма чотирма метриками має Random Forest: R² = 0.955, MAE ≈ 1 311 USD, MSE ≈ 3.52 млн, MAPE ≈ 9.6%.

Далі йдуть SVR і Gradient Boosting: R² = 0.934 і 0.931.

Лінійні моделі помітно слабші: R² 0.85–0.88, MAE 2.2–2.4 тис. USD, MAPE 16–18%. Залежність ціни від характеристик нелінійна і містить взаємодії (марка × двигун), які ансамблі дерев і RBF-ядро враховують, а лінійна модель — ні.

Ознаки перенавчання помірні. R² на train для ансамблів і SVR 0.985–0.996, на test 0.93–0.96.

## 12. Пояснення вибору типу моделі

Пояснити, чому класифікаційна модель, наприклад Logistic Regression, не підходить для безпосереднього прогнозування неперервної ціни автомобіля.

Зазначити, до якого типу задачі належить прогнозування `price` та чому моделі, використані в роботі, відповідають цій задачі.


Чому класифікація не підходить: Logistic Regression — модель класифікації. Вона оцінює ймовірність належності об'єкта до дискретних класів і повертає мітку. Ціна ж — неперервна величина. Модель класифікації не може спрогнозувати ціну, якої не було в train, і ігнорує відстань між значеннями.
Тип задачі: Прогнозування price — задача регресії. Усі використані моделі є регресорами, які повертають неперервний прогноз і мінімізують функцію втрат, що враховує величину відхилення (MSE, L1/L2, ε-нечутлива функція).

## 13. Візуальне порівняння прогнозів

Для кожної побудованої моделі зобразити співвідношення:
- фактичних значень `price`;
- прогнозованих значень `price`.

Додати лінію ідеального прогнозу та проаналізувати, наскільки близько до неї розташовані прогнозовані значення. Звернути увагу на великі відхилення.


```python
fig_scat, axes_scat = plt.subplots(2, 3, figsize=(18, 11))
val_lims = [0, max(te_y.max(), max(p.max() for p in preds_dict.values())) * 1.05]
text_offsets = [(-75, 8), (6, 8), (6, -18)]

for axis, m_name in zip(axes_scat.ravel(), df_metrics['Модель']):
    preds = preds_dict[m_name]
    abs_errors = np.abs(te_y.to_numpy() - preds)
    axis.scatter(te_y, preds, alpha=0.75, edgecolor='k', s=45)
    axis.plot(val_lims, val_lims, 'r--', lw=1.5, label='Ідеальний прогноз (y = x)')
    for idx_k, idx_j in enumerate(np.argsort(abs_errors)[-3:]):
        axis.annotate(f'ID {te_id.iloc[idx_j]}\nΔ={abs_errors[idx_j]:,.0f}', (te_y.iloc[idx_j], preds[idx_j]),
                    textcoords='offset points', xytext=text_offsets[idx_k], fontsize=8, color='darkred',
                    arrowprops=dict(arrowstyle='-', color='darkred', lw=0.6))
    row_stats = df_metrics.set_index('Модель').loc[m_name]
    axis.set_title(f'{m_name}\nR² = {row_stats["R²"]:.3f}, MAE = {row_stats["MAE"]:,.0f}, MAPE = {row_stats["MAPE, %"]:.1f}%')
    axis.set_xlim(val_lims)
    axis.set_ylim(val_lims)
    axis.set_xlabel('Фактична ціна, USD')
    axis.set_ylabel('Прогнозована ціна, USD')
    axis.legend(loc='upper left')
plt.suptitle('Фактичні та прогнозовані значення price (test)', fontsize=15)
plt.tight_layout()
plt.show()

max_errs_list = []
for m_name in df_metrics['Модель']:
    err_series = pd.Series(np.abs(te_y.to_numpy() - preds_dict[m_name]), index=te_id.to_numpy())
    top_errs = err_series.sort_values(ascending=False).head(3)
    max_errs_list.append({'Модель': m_name, 'car_ID з найбільшими помилками': list(top_errs.index),
                    'Абсолютні помилки': [round(v) for v in top_errs.values]})
display(pd.DataFrame(max_errs_list))
df_test_view = pd.DataFrame({'car_ID': te_id.to_numpy(), 'brand': te_X['brand'].to_numpy(),
                          'price': te_y.to_numpy()})
print('Найдорожчі автомобілі тестової вибірки:')
display(df_test_view.sort_values('price', ascending=False).head(5))
```

Найближче до лінії ідеального прогнозу точки Random Forest, SVR і Gradient Boosting. Для дешевих і середніх авто (до ~20 тис. USD) прогнози найточніші.

Найбільше відхилення — car_ID 17 (BMW, 41 315 USD): його недооцінюють усі моделі. Наявні ознаки моделі практично не розрізняють із дешевшими аналогами (наприклад, car_ID 16), тож частина помилки неусувна за цих даних.

SVR сильно переоцінює car_ID 19 (Chevrolet). Це єдине авто з категорією cylindernumber = 'three', якої немає в train.

Прокоментувати отримані результати.


```python
rf_model = pipes_dict['Random Forest']
rand_indices = np.random.RandomState(RAND_SEED).choice(len(te_X), size=10, replace=False)
sample_X = te_X.iloc[rand_indices]
sample_y = te_y.iloc[rand_indices]
sample_preds = rf_model.predict(sample_X)

df_sample_res = pd.DataFrame({
    'car_ID': te_id.iloc[rand_indices].to_numpy(),
    'Марка': sample_X['brand'].to_numpy(),
    'Фактична ціна': sample_y.to_numpy(),
    'Прогнозована ціна': sample_preds,
})
df_sample_res['Абсолютна помилка'] = (df_sample_res['Фактична ціна'] - df_sample_res['Прогнозована ціна']).abs()
df_sample_res['Абсолютна відсоткова помилка, %'] = (df_sample_res['Абсолютна помилка']
                                                   / df_sample_res['Фактична ціна'] * 100)
display(df_sample_res.style.format({'Фактична ціна': '{:,.0f}', 'Прогнозована ціна': '{:,.0f}',
                                   'Абсолютна помилка': '{:,.0f}',
                                   'Абсолютна відсоткова помилка, %': '{:.2f}'}))
print(f"Середня абсолютна помилка для 10 авто: {df_sample_res['Абсолютна помилка'].mean():,.0f} USD")
print(f"Середня абсолютна відсоткова помилка: {df_sample_res['Абсолютна відсоткова помилка, %'].mean():.2f}%")
print(f"Медіанна абсолютна відсоткова помилка: {df_sample_res['Абсолютна відсоткова помилка, %'].median():.2f}%")
print(f"Авто з помилкою < 10%: {(df_sample_res['Абсолютна відсоткова помилка, %'] < 10).sum()} з 10")
```

Для 10 випадкових авто з test середня абсолютна помилка Random Forest ≈ 1 329 USD, середня відсоткова ≈ 8.3% (медіана 7.2%).
Найточніші прогнози — для масових недорогих моделей (Toyota, Subaru, Mitsubishi), яких у train багато. Найбільші помилки виникають у рідкісних дорогих марок (наприклад, Volvo чи BMW), оскільки ансамбль дерев усереднює ціни подібних авто і не екстраполює за межі навчальних значень.
